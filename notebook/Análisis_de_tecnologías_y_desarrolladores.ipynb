# **4. Importación de librerías y carga de datos**

En esta sección se importan las librerías necesarias para realizar el análisis y se cargan los archivos de la encuesta. Pandas se utiliza para manipular los datos, NumPy para operaciones numéricas y Matplotlib para la creación de visualizaciones.

```python
# Dimensiones del dataset
filas, columnas = df.shape

print("Número de registros:", filas)
print("Número de columnas:", columnas)
df.head()
```

```python
print("Columnas disponibles:\n")

for columna in df.columns:
    print(columna)

df.info()
```

```python
# Cantidad de valores faltantes por columna
missing = df.isnull().sum()

missing = missing.sort_values(ascending=False)

missing.head(20)
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Valores_faltantes": df.isnull().sum(),
    "Porcentaje": missing_percent
})

missing_table = missing_table.sort_values(
    "Porcentaje",
    ascending=False
)

missing_table.head(20)
```

## 5. Selección de variables y preparación inicial

Se seleccionan las variables relacionadas con país, rol, experiencia, tecnologías y adopción de inteligencia artificial. A continuación, se revisan los valores faltantes y las categorías principales de las variables seleccionadas.

```python
# Buscar columnas relacionadas con tecnología, experiencia e IA

for columna in df.columns:
    if any(palabra in columna.lower() for palabra in [
        "language",
        "database",
        "ai",
        "years",
        "experience",
        "devtype",
        "country",
        "employment",
        "edlevel"
    ]):
        print(columna)

variables_interes = [
    "Country",
    "Employment",
    "DevType",
    "YearsCode",
    "EdLevel",
    "LanguageHaveWorkedWith",
    "DatabaseHaveWorkedWith",
    "AISelect"
]

for variable in variables_interes:
    if variable in df.columns:
        print("✅", variable)
    else:
        print("❌", variable, "NO encontrada")

df_analysis = df[variables_interes].copy()

print("Filas:", df_analysis.shape[0])
print("Columnas:", df_analysis.shape[1])



df_analysis.head()
df_analysis.info()
```

```python
missing_analysis = pd.DataFrame({
    "Valores_faltantes": df_analysis.isnull().sum(),
    "Porcentaje": df_analysis.isnull().mean() * 100
})

missing_analysis = missing_analysis.sort_values(
    "Porcentaje",
    ascending=False
)

missing_analysis


df_analysis["Employment"].value_counts().head(15)
df_analysis["DevType"].value_counts().head(15)
df_analysis["YearsCode"].value_counts().head(20)
df_analysis["EdLevel"].value_counts().head(15)
df_analysis["AISelect"].value_counts(dropna=False)

```

```python
df_analysis["LanguageHaveWorkedWith"].head(10)
df_analysis["DatabaseHaveWorkedWith"].head(10)
df_analysis["AISelect"].value_counts(dropna=False)
```

## 6. Análisis de tecnologías

Las preguntas de tecnologías permiten seleccionar varias respuestas. Por ello, se separan los valores de cada celda antes de contar las menciones. Una misma persona puede aparecer en varias tecnologías, así que los conteos no tienen por qué sumar el total de participantes.

```python
# Separar los lenguajes y contar cuántas veces aparece cada uno

languages = (
    df_analysis["LanguageHaveWorkedWith"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)

language_counts = languages.value_counts()

language_counts.head(15)

```

```python
import matplotlib.pyplot as plt

top_languages = language_counts.head(10).sort_values()

plt.figure(figsize=(10, 6))

top_languages.plot(kind="barh")

plt.title("Top 10 lenguajes de programación utilizados")
plt.xlabel("Número de desarrolladores")
plt.ylabel("Lenguaje")

plt.tight_layout()
plt.show()
```

```python
databases = (
    df_analysis["DatabaseHaveWorkedWith"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)

database_counts = databases.value_counts()

database_counts.head(15)
```

```python
top_databases = database_counts.head(10).sort_values()

plt.figure(figsize=(10, 6))

top_databases.plot(kind="barh")

plt.title("Top 10 bases de datos utilizadas")
plt.xlabel("Número de desarrolladores")
plt.ylabel("Base de datos")

plt.tight_layout()
plt.show()
```

```python
devtypes = (
    df_analysis["DevType"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)

devtype_counts = devtypes.value_counts()

devtype_counts.head(15)
```

```python
top_devtypes = devtype_counts.head(10).sort_values()

plt.figure(figsize=(10, 6))

top_devtypes.plot(kind="barh")

plt.title("Top 10 roles de desarrolladores")
plt.xlabel("Número de desarrolladores")
plt.ylabel("Rol")

plt.tight_layout()
plt.show()
```

## 7. Análisis de experiencia

La variable `YearsCode` contiene valores numéricos y algunas categorías escritas como texto. Se convierten estas categorías a aproximaciones numéricas y se agrupan los años de experiencia para facilitar la comparación.

```python
# Revisar los valores de experiencia
df_analysis["YearsCode"].value_counts(dropna=False).head(20)
```

```python
# Crear una copia de la variable original
df_analysis["YearsCode_num"] = df_analysis["YearsCode"].replace({
    "Less than 1 year": "0.5",
    "More than 50 years": "51"
})

# Convertir los valores a números
df_analysis["YearsCode_num"] = pd.to_numeric(
    df_analysis["YearsCode_num"],
    errors="coerce"
)

# Revisar el resultado
df_analysis[["YearsCode", "YearsCode_num"]].head(15)
```

```python
# Estadísticas de experiencia
df_analysis["YearsCode_num"].describe()
```

```python
# Crear grupos de experiencia
df_analysis["ExperienceGroup"] = pd.cut(
    df_analysis["YearsCode_num"],
    bins=[-1, 2, 5, float("inf")],
    labels=["Principiante", "Intermedio", "Avanzado"]
)

# Contar los desarrolladores por grupo
df_analysis["ExperienceGroup"].value_counts()
```

```python
import matplotlib.pyplot as plt

experience_counts = (
    df_analysis["ExperienceGroup"]
    .value_counts()
    .reindex(["Principiante", "Intermedio", "Avanzado"])
)

plt.figure(figsize=(8, 5))

experience_counts.plot(kind="bar")

plt.title("Desarrolladores según años de experiencia")
plt.xlabel("Grupo de experiencia")
plt.ylabel("Número de desarrolladores")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

## 8. Análisis de adopción de inteligencia artificial

Se examinan las respuestas de la variable `AISelect` y se comparan sus distribuciones entre los grupos de experiencia. Las diferencias observadas son descriptivas y no demuestran una relación causal.

```python
# Contar las respuestas sobre uso de IA
df_analysis["AISelect"].value_counts(dropna=False)
```

```python
# Calcular porcentajes de respuestas válidas
ai_counts = df_analysis["AISelect"].value_counts().dropna()

ai_percentages = (
    df_analysis["AISelect"]
    .value_counts(normalize=True)
    .dropna() * 100
)

ai_summary = pd.DataFrame({
    "Cantidad": ai_counts,
    "Porcentaje": ai_percentages
})

ai_summary
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

ai_counts.sort_values().plot(kind="barh")

plt.title("Respuestas sobre el uso de inteligencia artificial")
plt.xlabel("Número de desarrolladores")
plt.ylabel("Respuesta")

plt.tight_layout()
plt.show()
```

```python
# Tabla cruzada entre experiencia y uso de IA
ai_experience = pd.crosstab(
    df_analysis["ExperienceGroup"],
    df_analysis["AISelect"],
    normalize="index"
) * 100

ai_experience.round(2)
```

```python
ai_experience.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Uso de IA según grupo de experiencia")
plt.xlabel("Grupo de experiencia")
plt.ylabel("Porcentaje de respuestas")
plt.xticks(rotation=0)
plt.legend(title="Respuesta sobre IA", bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

## 9. Análisis geográfico

## 10. Resumen comparativo de resultados

Se presentan tablas de frecuencias y porcentajes para las tecnologías, junto con una comparación visual de los lenguajes y las bases de datos más mencionados.

```python
# Número de respuestas válidas sobre lenguajes
total_language_responses = df_analysis[
    "LanguageHaveWorkedWith"
].notna().sum()

# Tabla con cantidades y porcentajes
language_summary = pd.DataFrame({
    "Cantidad": language_counts.head(15),
    "Porcentaje": (
        language_counts.head(15) /
        total_language_responses * 100
    )
})

language_summary.round(2)
```

```python
total_database_responses = df_analysis[
    "DatabaseHaveWorkedWith"
].notna().sum()

database_summary = pd.DataFrame({
    "Cantidad": database_counts.head(15),
    "Porcentaje": (
        database_counts.head(15) /
        total_database_responses * 100
    )
})

database_summary.round(2)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

language_counts.head(5).sort_values().plot(
    kind="barh",
    ax=axes[0]
)
axes[0].set_title("Top 5 lenguajes")
axes[0].set_xlabel("Número de desarrolladores")
axes[0].set_ylabel("Lenguaje")

database_counts.head(5).sort_values().plot(
    kind="barh",
    ax=axes[1]
)
axes[1].set_title("Top 5 bases de datos")
axes[1].set_xlabel("Número de desarrolladores")
axes[1].set_ylabel("Base de datos")

plt.tight_layout()
plt.show()
```
